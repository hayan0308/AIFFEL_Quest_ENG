```python
import math


def get_integer(prompt):
    """정수가 입력될 때까지 반복해서 입력받는 함수"""
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            # 정수로 변환할 수 없으면 ValueError가 발생함
            print("잘못된 입력입니다. 정수를 입력해주세요.")


def calculate(num1, num2, operator):
    """연산자에 따라 계산을 수행하는 함수"""
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        # num2가 0이면 여기서 ZeroDivisionError가 발생함
        return num1 / num2
    elif operator == "**":
        # math 모듈을 사용한 제곱 연산 (결과는 실수형으로 반환됨)
        return math.pow(num1, num2)
    else:
        # 지원하지 않는 연산자
        return None


def main():
    print("=== 간단한 계산기 ===")

    while True:
        # 정수 두 개 입력받기 (정수가 들어올 때까지 반복)
        num1 = get_integer("첫 번째 정수를 입력하세요: ")
        num2 = get_integer("두 번째 정수를 입력하세요: ")

        # 연산자 입력받기
        operator = input("연산자를 입력하세요 (+, -, *, /, **): ")

        try:
            result = calculate(num1, num2, operator)
            if result is None:
                print("지원하지 않는 연산자입니다. (+, -, *, /, ** 중에서 입력해주세요.)")
            else:
                print(f"결과: {num1} {operator} {num2} = {result}")
        except ZeroDivisionError:
            print("0으로 나눌 수 없습니다.")

        # 계속 계산할지 물어보기
        again = input("계속 계산하시겠습니까? (y/n): ")
        if again.lower() != "y":
            print("계산기를 종료합니다.")
            break


if __name__ == "__main__":
    main()
```
