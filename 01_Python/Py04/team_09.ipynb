### **Q9.객체 순회 반복문을 통해 리스트에 있는 객체를 순회 하면서 잔고가 100    만원 이상인 고객의 정보만 출력하세요.**  

### **Q10입금과 출금내역이기록되도록코드를업데이트하세요.**  
(입금내역과출금내역을출력하는deposit_history 와withdraw_history  메서드를추가하세요.)


```python
import random

# =============================================
# 퀘스트 1번: 은행계좌 만들기
# =============================================

class Account:
    account_count = 0  # Q2. 클래스 변수: 생성된 계좌 수 저장

    def __init__(self, owner, balance):  # Q1. 생성자
        self.bank_name = "SC은행"
        self.owner = owner
        self.balance = balance
        self.account_number = self._generate_account_number()
        self._deposit_count = 0       # Q7. 입금 횟수 추적
        self._deposit_log = []        # Q10. 입금 내역
        self._withdraw_log = []       # Q10. 출금 내역
        Account.account_count += 1    # Q2. 계좌 생성 시 카운트 증가

    def _generate_account_number(self):
        """계좌번호 랜덤 생성 (111-11-111111 형태)"""
        part1 = str(random.randint(100, 999))
        part2 = str(random.randint(10, 99))
        part3 = str(random.randint(100000, 999999))
        return f"{part1}-{part2}-{part3}"

    @classmethod
    def get_account_num(cls):  # Q3. 클래스 변수 출력
        print(f"현재 생성된 계좌 수: {cls.account_count}개")

    def deposit(self, amount):  # Q4. 입금 메서드
        if amount < 1:
            print("입금은 최소 1원 이상만 가능합니다.")
            return
        self.balance += amount
        self._deposit_count += 1
        self._deposit_log.append(amount)  # Q10. 입금 내역 기록
        print(f"{amount:,}원이 입금되었습니다. 잔고: {self.balance:,}원")

        # Q7. 입금 5회마다 1% 이자 지급
        if self._deposit_count % 5 == 0:
            interest = int(self.balance * 0.01)
            self.balance += interest
            print(f"🎉 입금 5회 달성! 이자 {interest:,}원이 지급되었습니다. 잔고: {self.balance:,}원")

    def withdraw(self, amount):  # Q5. 출금 메서드
        if amount > self.balance:
            print("잔고가 부족합니다. 출금할 수 없습니다.")
            return
        self.balance -= amount
        self._withdraw_log.append(amount)  # Q10. 출금 내역 기록
        print(f"{amount:,}원이 출금되었습니다. 잔고: {self.balance:,}원")

    def display_info(self):  # Q6. 정보 출력 메서드
        print(f"은행이름: {self.bank_name}, 예금주: {self.owner}, "
              f"계좌번호: {self.account_number}, 잔고: {self.balance:,}원")

    def deposit_history(self):  # Q10. 입금 내역 출력
        print(f"[{self.owner}] 입금 내역:")
        if not self._deposit_log:
            print("  입금 내역이 없습니다.")
        for i, amount in enumerate(self._deposit_log, 1):
            print(f"  {i}회: +{amount:,}원")

    def withdraw_history(self):  # Q10. 출금 내역 출력
        print(f"[{self.owner}] 출금 내역:")
        if not self._withdraw_log:
            print("  출금 내역이 없습니다.")
        for i, amount in enumerate(self._withdraw_log, 1):
            print(f"  {i}회: -{amount:,}원")


# =============================================
# 테스트 코드
# =============================================

print("=" * 50)
print("퀘스트 1번: 은행계좌 만들기")
print("=" * 50)

# Q8. 3개 이상 인스턴스 생성 후 리스트에 저장
acc1 = Account("김철수", 500000)
acc2 = Account("이영희", 1500000)
acc3 = Account("박민준", 2000000)
acc4 = Account("최수연", 800000)

accounts = [acc1, acc2, acc3, acc4]

# Q3. 계좌 개수 출력
Account.get_account_num()

print()

# Q6. 정보 출력
print("[계좌 정보]")
for acc in accounts:
    acc.display_info()

print()

# Q4/Q5 입출금 테스트
print("[입출금 테스트 - 김철수 계좌]")
acc1.deposit(100000)
acc1.deposit(50000)
acc1.deposit(200000)
acc1.withdraw(30000)
acc1.deposit(10000)
acc1.deposit(20000)   # 5번째 입금 → 이자 지급!

print()

# Q10. 입출금 내역 출력
acc1.deposit_history()
acc1.withdraw_history()

print()

# Q9. 잔고 100만원 이상인 고객 정보 출력
print("[잔고 100만원 이상 고객]")
for acc in accounts:
    if acc.balance >= 1000000:
        acc.display_info()
```

● 생성된Monster  인스턴스와Player  인스턴스가battle   함수를통해전투  
○ player   는 생성된몬스터3마리(슬라임, 고블린, 오크)  와모두전투해야함  
○  전투도중에Player가 사망하면이후전투를진행하지않고 게임오버 출력하고종료  

```python
import random

# =============================================
# 퀘스트 2번: 간단한 자동사냥 RPG 만들기
# =============================================

# Q1. Character 클래스
class Character:
    def __init__(self, name, level, hp, attack, defense):
        self.name = name
        self.level = level
        self.hp = hp
        self.attack = attack
        self.defense = defense

    def is_alive(self):
        """현재 체력이 0 이상이면 True 반환"""
        return self.hp > 0

    def take_damage(self, damage):
        """데미지를 받으면 (데미지 - 방어력)만큼 체력 감소"""
        actual_damage = damage - self.defense
        if actual_damage > 0:
            self.hp -= actual_damage
        # 방어력이 데미지보다 크면 체력 감소 없음

    def attack_target(self, target):
        """타겟에게 1 ~ 공격력 사이의 랜덤 데미지를 입힘"""
        damage = random.randint(1, self.attack)
        target.take_damage(damage)
        print(f"{self.name}이(가) {target.name}에게 {damage}만큼 공격했다...!")
        print(f"{target.name}의 체력: {target.hp}")


# Q2. Player 클래스
class Player(Character):
    def __init__(self, name):
        super().__init__(
            name=name,
            level=1,
            hp=100,
            attack=25,
            defense=5
        )
        self.exp = 0  # 경험치 속성 추가

    def gain_experience(self, amount):
        """경험치 획득"""
        self.exp += amount
        print(f"{self.name}이(가) 경험치 {amount}를 획득했습니다! (현재 경험치: {self.exp})")

    def level_up(self):
        """경험치가 50 이상이면 레벨업"""
        while self.exp >= 50:
            self.exp -= 50
            self.level += 1
            self.attack += 10
            self.defense += 5
            print(f"🎉 레벨 업! {self.name}의 레벨이 {self.level}이 되었습니다! "
                  f"(공격력: {self.attack}, 방어력: {self.defense})")


# Q2. Monster 클래스
class Monster(Character):
    def __init__(self, name, level):
        hp = random.randint(10, 30) * level
        attack = random.randint(5, 20) * level
        defense = random.randint(1, 5) * level
        super().__init__(
            name=name,
            level=level,
            hp=hp,
            attack=attack,
            defense=defense
        )


# Q3. battle 함수
def battle(player, monster):
    """플레이어와 몬스터가 전투, 체력이 0 이하가 될 때까지 서로 공격"""
    print(f"\n{'='*40}")
    print(f"{monster.name} 과의 전투를 시작합니다.")
    print(f"{'='*40}")

    while player.is_alive() and monster.is_alive():
        player.attack_target(monster)   # 플레이어가 먼저 공격
        if not monster.is_alive():
            break
        monster.attack_target(player)   # 몬스터 반격

    if player.is_alive():
        exp_gain = monster.level * 20
        player.gain_experience(exp_gain)
        player.level_up()
        print("전투 승리!")
        print(f"{monster.name} 을(를) 이겼다!")
    else:
        print("전투 패배..")


# Q4. main 함수
def main():
    # 몬스터 딕셔너리
    monster_dict = {'슬라임': 1, '고블린': 2, '오크': 3}

    # 플레이어 생성
    player_name = input("플레이어 이름을 입력하세요: ")
    player = Player(player_name)
    print(f"\n{player.name} 님, 게임을 시작합니다! "
          f"(체력: {player.hp}, 공격력: {player.attack}, 방어력: {player.defense})\n")

    # 몬스터와 순서대로 전투
    for monster_name, monster_level in monster_dict.items():
        monster = Monster(monster_name, monster_level)
        battle(player, monster)

        if not player.is_alive():
            print("\n게임오버")
            return

    print(f"\n🏆 모든 몬스터를 물리쳤습니다! 최종 레벨: {player.level}")


if __name__ == "__main__":
    main()
```
