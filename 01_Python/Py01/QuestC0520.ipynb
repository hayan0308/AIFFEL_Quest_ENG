```python
'''
Q1-1 회문 판독기
'''
def is_palindrome(target):
    target_reverse= target[::-1]
    if target == target_reverse :
        return True
    else : 
        return False


target1 = "으아아"
target2 = "으아으"

print(is_palindrome(target1))
print(is_palindrome(target2))
```

```python
'''
Q1-2
'''
def palindrome_classifier(targets):
    #딕셔너리 정의 
    result={'회문O':[],'회문X':[]}

    for target in targets :
        if is_palindrome(target) == True:
            result['회문O'].append(target)
        else :
            result['회문X'].append(target)

    total = len(targets)
    palindrome = len(result['회문O'])
    # 일반적인 if-else 문으로 썼을 때
    if total > 0:
        ratio = palindrome / total
    else:
        ratio = 0.0

    return result, palindrome, ratio

targets1 = ['다시합시다', '기러기', '노트북', '일요일', '역삼역', '파이썬', '제주도']
print(palindrome_classifier(targets1))
    
```

```python
'''
Q2-1
'''

scores = {
    "유원영": {"국어": 94, "영어": 91, "수학": 89},
    "설지민": {"국어": 88, "영어": 96, "수학": 90},
    "신윤아": {"국어": 86, "영어": 88, "수학": 77},
    "장윤아": {"국어": 98, "영어": 76, "수학": 92}
}
```

```python
'''
Q2-2
'''
# 특정 과목의 학급 평균을 구하는 함수
def get_class_avg(subject):
    total = sum(student_scores[subject] for student_scores in scores.values())
    return total / len(scores)

# 메인 성적 분석 함수
def score_analysis(name):
    print(f"[{name} 성적 분석]")
    
    subjects = ["국어", "영어", "수학"]
    student_total = 0
    class_total_avg = 0
    
    # 1. 과목별 점수 출력 및 계산
    for sub in subjects:
        score = scores[name][sub]       # 학생의 과목 점수
        class_avg = get_class_avg(sub) # 해당 과목의 학급 평균
        diff = score - class_avg        # 대비 (차이)
        
        # 차이가 0 이상이면 + 기호를 붙여주고, 음수면 알아서 -가 붙음
        diff_str = f"+{diff:.2f}" if diff >= 0 else f"{diff:.2f}"
        
        print(f"{sub}: {score}점, 학급 평균({class_avg:.2f}점) 대비 {diff_str}")
        
        student_total += score
        class_total_avg += class_avg
    
    # 2. 전과목 평균 계산 및 출력
    student_avg = student_total / len(subjects)
    class_avg_all = class_total_avg / len(subjects)
    total_diff = student_avg - class_avg_all
    total_diff_str = f"+{total_diff:.2f}" if total_diff >= 0 else f"{total_diff:.2f}"

    print()
    print(f"전과목 평균: {student_avg:.2f}점, 학급 평균({class_avg_all:.2f}점) 대비 {total_diff_str}")

# 함수 실행 테스트
score_analysis("설지민")

```
